```python
import pandas as pd
```

```python
df=pd.read_csv("/content/breast-cancer-wisconsin-data - breast-cancer-wisconsin-data.csv")
```

```python
df
```

```python
x=df.drop(columns=["diagnosis","id"])
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

scaler = StandardScaler()
x_scaled_array = scaler.fit_transform(x)
x_scaled = pd.DataFrame(x_scaled_array, columns=x.columns)
display(x_scaled.head())
```

```python
y=df["diagnosis"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Initialize models
models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "Support Vector Classification": SVC(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42)
}

# Dictionary to hold the evaluation metrics
results = {}

# Train and evaluate each model
for name, model in models.items():
    # Fit the model
    model.fit(X_train, y_train)

    # Predict on test set
    y_pred = model.predict(X_test)

    # Calculate metrics (using pos_label='M' since 'M' stands for Malignant/positive class)
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, pos_label='M')
    recall = recall_score(y_test, y_pred, pos_label='M')
    f1 = f1_score(y_test, y_pred, pos_label='M')

    results[name] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    }

# Convert results to DataFrame for nice visualization
results_df = pd.DataFrame(results).T
display(results_df)
```

```python
from sklearn.model_selection import cross_validate
import pandas as pd
import numpy as np

# Define the scoring metrics we want to collect
# We define custom scorers or pass string identifiers. Since 'y' uses 'M' and 'B',
# we need to map them to binary 1 and 0 for metric functions to work automatically with cross_validate
y_binary = y.map({'M': 1, 'B': 0})

scoring = ['accuracy', 'precision', 'recall', 'f1']
cv_results_summary = {}

for name, model in models.items():
    # Run 5-Fold Stratified Cross-Validation
    cv_outputs = cross_validate(model, x_scaled, y_binary, cv=5, scoring=scoring)

    # Store mean scores
    cv_results_summary[name] = {
        "CV Accuracy (Mean)": np.mean(cv_outputs['test_accuracy']),
        "CV Precision (Mean)": np.mean(cv_outputs['test_precision']),
        "CV Recall (Mean)": np.mean(cv_outputs['test_recall']),
        "CV F1-Score (Mean)": np.mean(cv_outputs['test_f1'])
    }

# Convert results to DataFrame for nice visualization
cv_results_df = pd.DataFrame(cv_results_summary).T
display(cv_results_df)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Initialize the models
pca_models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "Support Vector Classification": SVC(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42)
}

pca_results = {}

# Train and evaluate each model on PCA features
for name, model in pca_models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    # Calculate metrics for the positive class 'M'
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, pos_label='M')
    recall = recall_score(y_test, y_pred, pos_label='M')
    f1 = f1_score(y_test, y_pred, pos_label='M')

    pca_results[name] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    }

# Display the results sorted by F1-Score and Accuracy
pca_results_df = pd.DataFrame(pca_results).T
display(pca_results_df.sort_values(by=["F1-Score", "Accuracy"], ascending=False))
```
